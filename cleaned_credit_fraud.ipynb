#Insepection of data

```python
import pandas as pd
import numpy as np

df = pd.read_csv("credit_fraud.csv")
df.info()
df.shape
df.head(10)
```

```python
#Checking null values and duplicates
```

```python
df.dropna(how = "all" , inplace = True)
print(df.isnull().sum())
print("total duplicate values are : ", df.duplicated().sum())
df = df.drop_duplicates()
```

#Filling the null values 

```python
#coverting age datatype
df['age'] = pd.to_numeric(df['age'], errors = "coerce")
df['age'] = df['age'].abs()
# filling null values
df = df.fillna({
    'age' : df['age'].median(),
    'transaction_amount' : df['transaction_amount'].median(),
    'account_balance' : df['account_balance'].median(),
    'num_transactions_today' : df['num_transactions_today'].median(),
    'is_foreign_transaction' : df['is_foreign_transaction'].mode()[0],
    'account_balance' : df['account_balance'].median(),
    'transaction_hour' : df['transaction_hour'].mode()[0],
    'prev_fraud_flag' : df['prev_fraud_flag'].mode()[0],
     'merchant_distance_km' : df['merchant_distance_km'].mean(),
     'merchant_risk_score' : df['merchant_risk_score'].mean()
         })
df.info()
```

#Changing datatypes

```python
int_type = ['age','num_transactions_today', 'transaction_hour', 'is_foreign_transaction', 'prev_fraud_flag','is_fraud']
df[int_type] = df[int_type].astype(int)
# df['age'] = df['age'].astype(int)
df.info()
```

#Binary to string

```python
def yes_no(value):
    if value == 1:
        return "yes"
    else:
        return "no"
# df[['is_foreign_transaction', 'prev_fraud_flag', 'is_fraud']] = df[['is_foreign_transaction', 'prev_fraud_flag', 'is_fraud']].apply(yes_no)
df['is_foreign_transaction'] = df['is_foreign_transaction'].apply(yes_no)
df['prev_fraud_flag'] = df['prev_fraud_flag'].apply(yes_no)
df['is_fraud'] = df['is_fraud'].apply(yes_no)
df.head(10)
df.to_csv("cleaned_credit_fruad_data.csv", index = False)
```
