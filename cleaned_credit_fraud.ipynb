# Importing the libraries

```python
import pandas as pd
import numpy as np
```

# Loading the dataset

```python
df = pd.read_csv("credit_fraud.csv")
```

# Inspection of the dataset

```python
df.info()
df.head(10)
```

# Removing empty rows and duplicates

```python
df.dropna(how = "all" , inplace = True)
df = df.drop_duplicates()
```

# Checking null values

```python
print(df.isnull().sum())
```

# Removing null values of rows

```python
df = df.dropna(thresh = 5)
```

# Fixing null values

```python
df['age'] = pd.to_numeric(df['age'], errors = "coerce")


df = df.fillna({
    'age' : df['age'].median(),
    'transaction_amount' : df['transaction_amount'].median(),
    'account_balance' : df['account_balance'].median(),
    'num_transactions_today' : df['num_transactions_today'].median(),
    'is_foreign_transaction' : df['is_foreign_transaction'].mode()[0],
    'account_balance' : df['account_balance'].median(),
    'transaction_hour' : df['transaction_hour'].median(),
    'prev_fraud_flag' : df['prev_fraud_flag'].mode()[0],
     'merchant_distance_km' : df['merchant_distance_km'].mean(),
     'merchant_risk_score' : df['merchant_risk_score'].mean()
         })
```

```python
df.info()
```

# coverting datatypes

```python
int_type = ['age','num_transactions_today', 'transaction_hour', 'is_foreign_transaction', 'prev_fraud_flag','is_fraud']
df[int_type] = df[int_type].astype(int)

binary_cols = ['is_foreign_transaction','prev_fraud_flag', 'is_fraud']
for cols in binary_cols:
    df[cols] = df[cols].map({1: "yes", 0 :"no"})
```

# Checking Outliers

```python
df.describe()
```

# Removing Outliers

```python
df = df[df['transaction_hour'] <= 23]

df = df[df['account_balance'] > 100]

q1 = df['transaction_amount'].quantile(0.25)
q3 = df['transaction_amount'].quantile(0.75)
iqr = q3 - q1
lower = q1 - (1.5 * iqr)
upper = q3 + (1.5 * iqr)
df = df[(df['transaction_amount'] >= lower) & (df['transaction_amount'] <= upper)]
```
